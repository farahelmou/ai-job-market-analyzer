Phase 2 : Data Cleaning 

```python
import pandas as pd
```

Step 1 — Work on the merged DataFrame

```python
df=pd.read_csv("../Data/processed/Ai_jobs.csv")
df.info()
df.head()
```

1-Missing values

```python
(df.isnull().sum() / len(df)) * 100
```

```python
df=df.drop(columns=["Unnamed: 0"])
df["skills required"]=df["skills required"].fillna("Unknown")
df["is fresher?"]=df["is fresher?"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

2-check duplicates

```python
df.duplicated().sum()
```

3-Check inconsistent values

```python
df["is remote?"].value_counts()
```

```python
df["is fresher?"].value_counts()
```

As two different values, so we'd standardize them

```python
df["is remote?"]=df["is remote?"].str.strip().str.lower()
```

```python
df["is fresher?"]=df["is fresher?"].str.strip().str.lower()
```

4-Clean text Columns

```python
text_columns=["job title","company","region","skills required"]
for col in text_columns:
    df[col]=df[col].str.strip()
```
