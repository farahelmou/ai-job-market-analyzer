Phase 1 : Data exploration 

```python
import pandas as pd
```

Exploring Data

```python
df_data_analyst=pd.read_csv("../Data/raw/Data_Analyst_jobs.csv")
df_data_science_jobs=pd.read_csv("../Data/raw/Data_science_jobs.csv")
df_ml_engineer_jobs=pd.read_csv("../Data/raw/ML_Engineer_jobs.csv")
```

Merging them by stacking the rows vertically 

```python
df_data_analyst["job_category"]="Data analyst"
df_data_science_jobs["job_category"]="Data scientist"
df_ml_engineer_jobs["job_category"]="ML engineer"
df=pd.concat([df_data_analyst,df_data_science_jobs,df_ml_engineer_jobs], ignore_index=True)
print(df.shape)
print(df.head())
```
